## Multiple Linear Regression

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# 1. Loading data

df = pd.read_csv('./data.csv')
df
```

```python
# 2. train the model

model = LinearRegression()

X = df[['Experience_Years','Training_Hours']]
Y = df['Performance_Score']

model.fit(X,Y)
```

```python
# 3-D Plot

fig = plt.figure(figsize=(10,8))
ax = fig.add_subplot(111, projection='3d')

# Actual poits
ax.scatter(df['Experience_Years'],df['Training_Hours'],df['Performance_Score'])

# Create meshgrid for plane
x_surf, y_surf = np.meshgrid(
  np.linspace(df['Experience_Years'].min(),df['Experience_Years'].max(),20),
  np.linspace(df['Training_Hours'].min(),df['Training_Hours'].max(),20)
)

# Predict over the grid
z_surf = model.predict(np.array([x_surf.ravel(), y_surf.ravel()]).T)
z_surf = z_surf.reshape(x_surf.shape)


# Surface plot
ax.plot_surface(x_surf, y_surf, z_surf, alpha=0.5)

ax.set_xlabel('Experience_Years')
ax.set_ylabel('Training_Hours')
ax.set_zlabel('Performance_Score')
plt.show()
plt.show()
```
